```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b"
)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant that helps to create and curate posts for a social media platform."),
    ("human","I want to create a post about {topic}")
])
```

```python
chain = prompt | llm
```

```python
res = chain.invoke({"topic": "The importance of data privacy in the digital age"})
res.content
```
